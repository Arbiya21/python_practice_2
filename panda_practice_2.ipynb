```python
import pandas as pd
```

```python
data = {
    'Name': ['Arbiya', 'Ayesha', 'Sara', 'Arbiya', None],
    'Age': [22, 20, 21, 22, None],
    'Salary': [20000, 25000, 30000, 20000, 35000]
}
```

```python
df = pd.DataFrame(data)
```

```python
print(df)
```

```python
df.dropna()
```

```python
df.fillna(0)
```

```python
df['Name'] = df['Name'].fillna('Unknown')
```

```python
df.drop_duplicates()
```

```python
df.duplicated()
```

```python
df['Name'] = df['Name'].replace('Arbiya', 'Arbiya E')
```

```python
df.rename(columns={'Name': 'Student_Name'})
```

```python
df.drop(columns=['Salary'])
```
